#  Data Inspection
Data inspection helps us understand the dataset's size, column names, data types, missing information, unique values, and potential target variables.

```python
import pandas as pd
import numpy as np

# Load the Online Retail dataset
df = pd.read_csv("data.csv", encoding="cp1252")

print("Dataset loaded successfully!")
```

## Number of rows
Rows represent records

```python
print("Number of rows:", df.shape[0])
print("Dataset dimensions:", df.shape)
```

### Number of  columns
 Columns represent attributes of each record.

```python

print("Number of columns:", df.shape[1])
print("Dataset dimensions:", df.shape)
```

## Column names
Column names tell us what information is available in the dataset.

```python
print(df.columns)
print("\nColumn names as a list:")
print(df.columns.tolist())
```

## Index
The index identifies the rows in a Pandas DataFrame. By default, Pandas assigns integer labels starting from zero.

```python
print(df.index)
print("\nFirst 10 index values:")
print(df.index[:10])
```

## head() — View the first rows
The head() method displays the first five rows by default.

```python
df.head()
```

## tail() 
View the last rows

```python
# Display the last five rows
df.tail()

# Display the last 10 rows
df.tail(10)
```

## sample() 
View random rows

```python
# Display five random rows
df.sample(5, random_state=42)
```

## shape 
Dataset dimensions

```python
rows, columns = df.shape

print("Rows:", rows)
print("Columns:", columns)
print("Total cells:", df.size)
```

## info() 
Dataset information

```python
df.info()
```

## describe() 
Statistical summary

```python
# Statistical summary of numerical columns
df.describe()
```

## dtypes 
Check data types

```python
print(df.dtypes)
```

## nunique() 
Count unique values
The nunique() method counts distinct, non-missing values in each column by default.

```python
print(df.nunique())
```

## value_counts() 
Count values in a column
This method counts how frequently each value occurs.


```python
print(df["Country"].value_counts())
```
