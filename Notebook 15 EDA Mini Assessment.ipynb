## What is EDA?
EDA stands for Exploratory Data Analysis.

EDA is the process of analysing a dataset to understand its structure, identify problems, discover patterns and understand relationships between variables before building a Machine Learning model.

## Why is EDA performed before Machine Learning?
EDA (Exploratory Data Analysis) is performed before Machine Learning to understand the dataset, identify problems, discover patterns and prepare clean, meaningful data for building an accurate Machine Learning model.
### Main reasons for performing EDA
1. Understand the dataset: Identify the number of rows, columns, features and data types.
2. Identify missing values: Find empty or unavailable data and decide how to handle it.
3. Detect duplicate records: Identify repeated data that may affect model performance.
4. Find outliers: Identify unusual values and investigate whether they are genuine or incorrect.
5. Understand data distribution: Check whether numerical data is normally distributed, positively skewed or negatively skewed.
6. Identify relationships: Understand how different features are related using correlation.
7. Detect class imbalance: Check whether one target category has significantly more records than another.
8. Select useful features: Identify which features may help the Machine Learning model make predictions.

```python
import pandas as pd

data = {
    "Study_Hours": [2, 4, 6, 8, 5],
    "Attendance": [50, 65, 80, 90, None],
    "Previous_Marks": [40, 55, 70, 85, 60],
    "Final_Marks": [35, 50, 75, 90, 65]
}

df = pd.DataFrame(data)

# Understand dataset
print(df.head())

# Check missing values
print(df.isnull().sum())

# Check duplicate records
print(df.duplicated().sum())

# Statistical analysis
print(df.describe())

# Correlation analysis
print(df.corr())
```

## Difference Between Univariate, Bivariate, and Multivariate Analysis
In Exploratory Data Analysis (EDA), we use three important types of analysis to understand a dataset:
1. Univariate Analysis
2. Bivariate Analysis
3. Multivariate Analysis
   
The main difference is the number of variables we analyse at a time.
#### 1. Univariate Analysis
Definition: Univariate analysis is the process of analysing only one variable at a time to understand its distribution, frequency and statistical properties.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Marks": [45, 60, 75, 85, 90]
}

df = pd.DataFrame(data)

# Statistical analysis
print(df["Marks"].describe())

# Mean
print("Mean:", df["Marks"].mean())

# Median
print("Median:", df["Marks"].median())

# Histogram
sns.histplot(df["Marks"], bins=5)

plt.title("Marks Distribution")
plt.show()
```

### Bivariate Analysis
Definition: Bivariate analysis is the process of analysing two variables together to understand the relationship or association between them.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Study_Hours": [2, 3, 5, 7, 9],
    "Marks": [40, 50, 65, 80, 95]
}

df = pd.DataFrame(data)

# Calculate correlation
print(
    df["Study_Hours"].corr(df["Marks"])
)

# Scatter plot
sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Marks"
)

plt.title("Study Hours vs Marks")
plt.show()
```

## How do you identify Duplicate Records in a Dataset?
#### What are Duplicate Records?
Duplicate records are repeated rows or entries in a dataset that contain the same information.

In simple words, when the same record appears more than once in a dataset, it is called a duplicate record.
#### Why do duplicate records occur?
Duplicate records can occur for several reasons:
1. Data entry errors: A user enters the same information multiple times.
2. Data collection errors: The same information is collected repeatedly.
3. Import errors: Records are duplicated while importing CSV or Excel files.
4. System errors: Software accidentally saves the same record more than once.
5. Data merging: Combining multiple datasets may introduce duplicate records.

## What is an Outlier in Data Analysis?
An outlier is a data value that is very different from the other values in a dataset.

In simple words, an outlier is an unusually high or unusually low value compared with the remaining data.

Outliers are identified during Exploratory Data Analysis (EDA) because they can affect statistical calculations and Machine Learning models.
#### Types of outliers
There are three common types:

1. High-value outlier

A value that is unusually higher than the other values.

Example: Employee salaries = ₹20,000, ₹22,000, ₹25,000, ₹28,000, ₹5,00,000.

2. Low-value outlier

A value that is unusually lower than the other values.

Example: Student ages = 18, 19, 20, 21, 2.

3. Contextual outlier

A value that is unusual only in a particular situation.

Example: A temperature of 40°C may be normal in summer but unusual during winter in some locations.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    'Marks': [45, 50, 55, 60, 98]
}

df = pd.DataFrame(data)

sns.boxplot(x=df['Marks'])
plt.title("Outlier Detection Using Box Plot")
plt.show()
```

## What is IQR?
IQR stands for Interquartile Range.

It measures the spread of the middle 50% of the data.

To calculate IQR, we need two important values:
- Q1 (First Quartile): The value below which 25% of the data falls.
- Q3 (Third Quartile): The value below which 75% of the data falls.
   #### Formula:
  
  

```python
import pandas as pd

# Create dataset
data = {
    'Marks': [10, 20, 30, 40, 50, 60, 70, 80, 300]
}

df = pd.DataFrame(data)

# Calculate Q1 and Q3
Q1 = df['Marks'].quantile(0.25)
Q3 = df['Marks'].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Identify potential outliers
outliers = df[
    (df['Marks'] < lower_limit) |
    (df['Marks'] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print("Potential Outliers:")
print(outliers)
```

## Detect Outliers Using IQR Method
IQR (Interquartile Range) is a statistical method used to identify outliers in a dataset.

It measures the spread of the middle 50% of the data and helps detect unusually low or high values.
1. Understand the IQR Formula
   


- Q1 (First Quartile): 25% of the data falls below this value.
- Q3 (Third Quartile): 75% of the data falls below this value.
- IQR: Difference between Q3 and Q1.

```python
import pandas as pd
import numpy as np

# Select numerical columns
numerical_columns = df.select_dtypes(include=np.number).columns

# Detect outliers using IQR
for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\nColumn: {column}")
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))
```

## Generate a Complete EDA Summary
A Complete EDA (Exploratory Data Analysis) Summary is a final report that combines all the findings from dataset exploration.

It helps us understand the dataset's structure, quality, patterns, relationships, and potential problems before building a Machine Learning model.

```python
import pandas as pd
import numpy as np

# ==================================
# 1. DATASET OVERVIEW
# ==================================

print("=" * 50)
print("           COMPLETE EDA SUMMARY")
print("=" * 50)

print("\n1. DATASET OVERVIEW")

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()


# ==================================
# 2. MISSING VALUE ANALYSIS
# ==================================

print("\n2. MISSING VALUE ANALYSIS")

missing = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing Percentage': (
        df.isnull().mean() * 100
    )
})

print(missing)


# ==================================
# 3. DUPLICATE ANALYSIS
# ==================================

print("\n3. DUPLICATE ANALYSIS")

print("Duplicate Records:", df.duplicated().sum())


# ==================================
# 4. NUMERICAL ANALYSIS
# ==================================

print("\n4. NUMERICAL ANALYSIS")

numerical_columns = df.select_dtypes(
    include=np.number
).columns

print(df[numerical_columns].describe())


# ==================================
# 5. CATEGORICAL ANALYSIS
# ==================================

print("\n5. CATEGORICAL ANALYSIS")

categorical_columns = df.select_dtypes(
    include=['object', 'category', 'bool']
).columns

for column in categorical_columns:

    print(f"\nColumn: {column}")

    print("Unique Values:", df[column].nunique())

    print("Category Counts:")
    print(df[column].value_counts())


# ==================================
# 6. OUTLIER ANALYSIS USING IQR
# ==================================

print("\n6. OUTLIER ANALYSIS")

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print(column, "Outliers:", len(outliers))


# ==================================
# 7. DISTRIBUTION ANALYSIS
# ==================================

print("\n7. DISTRIBUTION ANALYSIS")

for column in numerical_columns:

    print(f"\nFeature: {column}")

    print("Mean:", df[column].mean())
    print("Median:", df[column].median())
    print("Skewness:", df[column].skew())
    print("Kurtosis:", df[column].kurt())


# ==================================
# 8. CORRELATION ANALYSIS
# ==================================

print("\n8. CORRELATION ANALYSIS")

correlation = df[numerical_columns].corr()

print(correlation)


# ==================================
# 9. HIGH CORRELATION CHECK
# ==================================

print("\n9. HIGH CORRELATION FEATURES")

corr_pairs = correlation.where(
    np.triu(np.ones(correlation.shape), k=1).astype(bool)
).stack()

print(corr_pairs[abs(corr_pairs) > 0.8])


# ==================================
# 10. CLASS DISTRIBUTION
# ==================================

print("\n10. TARGET CLASS DISTRIBUTION")

target_column = 'target'

if target_column in df.columns:

    print(df[target_column].value_counts())

    print("\nClass Percentages:")

    print(
        df[target_column].value_counts(
            normalize=True
        ) * 100
    )

else:

    print("Target column not found. Update target_column.")


print("\n" + "=" * 50)
print("           EDA COMPLETED")
print("=" * 50)
```
