## What is a Dataset?
A dataset is a collection of related information organized into rows and columns.
### Example:
In your retail dataset:
- Each row contains information about a product on an invoice.
- Each column represents an attribute of that transaction.
- The entire CSV file is the dataset.

```python
# Display five random records
df.sample(5, random_state=42)
```

## What is a Feature?
A feature is a column containing information that can help us analyze data or make predictions.
### Examples:
- Quantity — how many units were purchased or recorded.
- UnitPrice — the price per unit.
- Country — the customer's country.
- InvoiceDate — when the transaction occurred.
- StockCode — the product identifier.

```python
# Display all features (columns)
features = df.columns.tolist()

print("Features in the dataset:")
for feature in features:
    print(feature)
```

## What is a Target?
A target is the variable that a machine learning model is designed to predict.



```python
# Select the target variable
target = "Quantity"

print("Target variable:", target)
print(df[target].head())
```

## Independent variables (X)
These are the input variables used by a model to predict the target.
## Dependent variable (y)
This is the output variable we want the model to predict.
For our example, predicting Quantity:
- Independent variables: StockCode, UnitPrice, Country, and suitable date features.
- Dependent variable: Quantity

```python
# Define the target
y = df["Quantity"]

# Define example input features
X = df[["StockCode", "UnitPrice", "Country"]]

print("Input features (X):")
print(X.head())

print("\nTarget variable (y):")
print(y.head())
```

## Numerical features
Numerical features contain numeric values that can be measured or counted.
### Example:
- Quantity — integer counts.
- UnitPrice — decimal-valued prices.
- CustomerID — appears numeric, but it is an identifier, not a measurement.

## Categorical features
Categorical features represent groups, labels, or categories.
### Example:
- Country — country category.
- Description — product name or description.
- StockCode — product code, which functions as an identifier/category rather than a numerical measurement.
- InvoiceNo — invoice identifier.

```python
# Automatically identify numerical and non-numerical columns
numerical_columns = df.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("Numerical columns:", numerical_columns)
print("Categorical/text columns:", categorical_columns)
```

## Discrete features
Discrete features contain countable values, often whole numbers.
### Example: 
Quantity counts units of a product.

```python
print("Quantity data type:", df["Quantity"].dtype)
print("Example quantities:")
print(df["Quantity"].head(10).tolist())

print("Unique quantity values:")
print(df["Quantity"].nunique())
```

## Continuous features
Continuous features represent measurements that can take decimal values within a range.
### Example: 
    UnitPrice can contain values such as 2.55 and 3.39.

```python
print("UnitPrice data type:", df["UnitPrice"].dtype)
print("Example prices:")
print(df["UnitPrice"].head(10).tolist())

print("Minimum price:", df["UnitPrice"].min())
print("Maximum price:", df["UnitPrice"].max())
```

## Identifier columns
Identifier columns distinguish records, invoices, products, or customers. They do not necessarily measure a quantity.
### Example:
- InvoiceNo — invoice or transaction identifier.
- StockCode — product identifier.
- CustomerID — customer identifier.

```python
identifier_columns = [
    "InvoiceNo",
    "StockCode",
    "CustomerID"
]

for col in identifier_columns:
    if col in df.columns:
        print(f"\n{col}")
        print("Unique values:", df[col].nunique())
        print("Missing values:", df[col].isnull().sum())
```

## Target variable
Metadata is information that describes the dataset, its structure, or the meaning of its fields. Some metadata exists outside the CSV, such as the dataset source, collection period, and definitions of columns.

```python
column_descriptions = {
    "InvoiceNo": "Invoice or transaction identifier",
    "StockCode": "Product identifier",
    "Description": "Product description",
    "Quantity": "Number of units recorded",
    "InvoiceDate": "Date and time of the transaction",
    "UnitPrice": "Price per unit",
    "CustomerID": "Customer identifier",
    "Country": "Country associated with the transaction"
}

for column, description in column_descriptions.items():
    print(f"{column}: {description}")
```
